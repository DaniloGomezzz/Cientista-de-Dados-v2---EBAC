```python
df.info()
```

```python
contagem_genero = df['Genero'].value_counts()

porcentagem_genero = (contagem_genero / contagem_genero.sum()) * 100

ax = contagem_genero.plot(kind='bar')
plt.title("Distribuição de Gênero")
plt.xlabel("Gênero")
plt.ylabel("Quantidade")
```

```python
contagem_idoso = df['Idoso'].value_counts()

porcentagem_idoso = (contagem_idoso / contagem_idoso.sum()) * 100

ax = contagem_idoso.plot(kind='bar')
plt.title("Distribuição de Idoso")
plt.xlabel("Idoso")
plt.ylabel("Quantidade")
```

```python
contagem_dependentes = df['Tipo_Contrato'].value_counts()

porcentagem_dependentes = (contagem_dependentes / contagem_dependentes.sum()) * 100

ax = contagem_dependentes.plot(kind='bar')
plt.title("Distribuição de Tipos de Contrato")
plt.xlabel("Tipo de Contrato")
plt.ylabel("Quantidade")
```

```python
contagem_churn = df['Churn'].value_counts()

porcentagem_churn = (contagem_churn / contagem_churn.sum()) * 100

ax = contagem_churn.plot(kind='bar')
plt.title("Distribuição de Churn")
plt.xlabel("Churn")
plt.ylabel("Quantidade")
```

# **D)**

A distribuição de Churn está boa, poucos clientes são perdidos em comparação aos que sã mantidos.

Há uma esquivalência entre os gêneros, então está bem balanceado também.

# **A)**

```python
fig = px.histogram(df, x='Churn', color='Genero', barmode='stack')

fig.update_layout(title_text='Relação: Churn x Gênero',
                  xaxis_title='Churn',
                  yaxis_title='Contagem',
                  legend_title='Gênero')

fig.show()
```

```python
fig = px.histogram(df, x='Churn', color='Idoso', barmode='stack')

fig.update_layout(title_text='Relação: Churn x Idoso',
                  xaxis_title='Churn',
                  yaxis_title='Contagem',
                  legend_title='Idoso')

fig.show()
```

```python
media_pagamento_mensal = df.groupby('Churn')['Pagamento_Mensal'].median().reset_index()

fig = px.bar(media_pagamento_mensal, x='Churn', y='Pagamento_Mensal',
             labels={'Churn': 'Churn', 'Pagamento_Mensal': 'Média do Pagamento Mensal'},
             title='Média do Pagamento Mensal por Churn')

fig.show()
```

```python
fig = px.histogram(df, x='Churn', color='Tipo_Contrato', barmode='stack')

fig.update_layout(title_text='Relação: Churn x Tipo de Contrato',
                  xaxis_title='Churn',
                  yaxis_title='Contagem',
                  legend_title='Tipo de Contrato')

fig.show()
```

# **B)**

A partir dos gráficos apresentados acima, podemos inferir aluns insights:



*   Distribuição de Gênero por Churn é equivalente, ou seja, há uma distribuição semelhante entre Gênero e Churn;

*   Para idosos, também é indiferente. Uma pssibilidade seria se a maior quantidade de idoso preferisse abandonar o contrato, mas não é o caso. Os valores para esse caso também são equivalentes e não apresentam problemas;

*   O Pagamento Mensal também não aparenta ser um problema. Tiramos a média entre Churn 'Yes' e Churn 'No' para compará-los, porém desmonstrou não ser um problema também. A média e a quantidade de Churn são semelhantes e o valor Mensal Pago não parece ser o causador da perda de clientes;

*   Enfim, temos então a análise na coluna Tipo de Contrato, em que temos 3 tipos: mês-a-mês, a cada ano, e a cada dois anos. Nesse caso conseguimos notar uma discrepância enorme entre os contratos quando observamos a coluna 'Churn Yes'. O que nos mostra que os clientes com contrato Mês-a-Mês são os mais propensos a deixarem de ser clientes, enquanto que para os outros contratos tempos pouquíssimas desistências.
