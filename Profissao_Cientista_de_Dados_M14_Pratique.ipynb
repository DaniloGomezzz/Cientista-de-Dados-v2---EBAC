```python
sns.set_style('darkgrid')

plt.figure(figsize=(10,6))
sns.histplot(data=df, x='Pagamento_Mensal', bins=20, kde=True, color='skyblue')
plt.xlabel('Pagamento Mensal')
plt.ylabel('Frequência')
plt.title('Distribuição do Pagamento Mensal')
plt.show()
```

```python
#seu código aqui

mediana = df['Pagamento_Mensal'].median()

df['Pagamento_Mensal'].fillna(mediana, inplace=True)
```
